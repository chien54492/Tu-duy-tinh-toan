Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu mỗi phần tử x trong mảng xuất hiện đúng x lần. Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.

```python
n=list(map(int,input().split()))
t=0
d={}
for i in n:
    if i in d:
        d[i]+=1
    else:
        d[i]=1
for key,value in d.items():
    
    if key==value:
        continue
    if value>key:
        t+=value-key
    else:
        t+=value
print(t)

```

W7A27
Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ hay không

```python
s = input()
st = []
d = {
    ')': '(',
    '}': '{',
    ']': '['}
ok = True
for x in s:
    if x in "({[":
        st.append(x)
    else:
        if len(st) == 0:
            ok = False
            break
        if st[-1] != d[x]:
            ok = False
            break
        st.pop()
if len(st) != 0:
    ok = False
if ok:
    print("YES")
else:
    print("NO")
```

W7A28
Bà̀i toán Tìm dãy các hợp số dài nhất liên tiếp không chứa số nguyên nào <= N (N là số nguyên dương cho trước). Nếu có nhiều dãy cùng độ dài thì lấy dãy nhỏ nhất (gặp đầu tiên từ trái qua)

```python
import math
def hop_so(x):
    if x <= 3:
        return False
    for i in range(2, int(math.sqrt(x)) + 1):
        if x % i == 0:
            return True
    return False
n = int(input())
t = []
hs = []
for i in range(1, n + 1):
    if hop_so(i):
        hs.append(i)

    else:
        if len(hs) > len(t):
            t = hs[:]

        hs = []

if len(hs) > len(t):
    t = hs[:]

print(*t)
```

W7A29
Tìm Dãy con tăng dần liên tục dài nhất của một danh sách nhập vào từ bàn phím. Lấy dãy đầu tiên nếu có nhiều dãy có số phần tử bằng nhau

```python
a = list(map(int, input().split()))
t = []
c = [a[0]]

for i in range(1, len(a)):

    if a[i] > a[i-1]:
        c.append(a[i])

    else:
        if len(c) > len(t):
            t = c[:]

        c = [a[i]]

if len(c) > len(t):
    t = c[:]

print(*t)
```
