Viết hàm tìm kiếm nhị phân trên một danh sách đã được sắp xếp tăng dần. Trả về chỉ số của phần tử hoặc -1 nếu không tìm thấy. mẫu hàm: def binary_search(arr, left, right, target):

```python
def binary_search(arr, target):
    arr.sort()
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1
arr=list(map(int, input().split()))
target=int(input())
print(binary_search(arr, target))
    
```

W7A8
Bạn được cung cấp: một danh sách các khoảng thời gian interval = [[l1, r1], [l2, r2], ...] và một danh sách các truy vấn queries = [q1, q2, ...] Với mỗi truy vấn q, hãy tìm độ dài nhỏ nhất của khoảng thời gian [li, ri] sao cho li ≤ q ≤ ri. Nếu không tồn tại khoảng thời gian nào như vậy, hãy trả về -1 cho truy vấn đó. solve in < O(m.n), hint: binary search in the queries list

```python
lt=eval(input())
d=list(map(int, input().split()))
n=[]
for q in d:
    m=[]
    for l,r in lt:
        if l<=q and q<=r:
            k=r-l+1
            m.append(k)
    if len(m) == 0:
        n.append(-1)
    else:
        n.append(min(m)) 

print(n)

```

Cho một danh sách các số
nums, với mỗi phần tử
nums[i], hãy đếm xem có
bao nhiêu phần tử bên phải
nhỏ hơn nums[i].

đếm.
Efficient solution: using
heap or binary search tree

```python
n=list(map(int, input().split()))
t=[]
for i in range(len(n)):
    s=0
    for j in range(i+1, len(n)):
        if n[j] < n[i]:
            s+=1
    t.append(s)
print(t)
```

Dãy hài hòa là một dãy
thỏa mãn điều kiện giá trị
lớn nhất và giá trị nhỏ nhất
của dãy khác nhau đúng
bằng \( 1 \).
Viết hàm trả về độ dài của
dãy hài hòa con dài nhất có
thể có trong mảng giá trị
nguyên \( arr \) có \( n \)
phần tử
Ví dụ:
Input: 1 3 2 2 5 2 3 7
Output: 5 - Dãy hài hòa con
dài nhất là [3,2,2,2,3]

```python
n=list(map(int, input().split()))
t=[]
s=[]
for i in range(len(n)-1):
    for j in range(i+1, len(n)):
       if n[j]-n[i]==1:
        s.append(n[i])
        s.append(n[j])
    t.append(s)
    h=max(t, key=len)
print(h)


```

W7A14
Hãy viết một chương
trình tìm kiếm tất cả các số
7 có trong một dãy số.
Đầu vào: số nguyên n trên
một dòng và n số nguyên ở
dòng kế tiếp.
Đầu ra:
Vị trí của các số \( 7 \)
trong dãy cách nhau bởi
dấu cách, vị trí lớn hơn
đứng trước và bắt đầu tính
từ \(0\).
Nếu không tìm được số \( 7
\) nào in ra "Not found" 

```python
n = int(input())
a = list(map(int, input().split()))

res = []

for i in range(n):
    if a[i] == 7:
        res.append(i)

res.reverse()

if len(res) == 0:
    print("Not found")
else:
    print(*res)
```
