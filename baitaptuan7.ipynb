Viết hàm tìm kiếm nhị phân trên một danh sách đã được sắp xếp tăng dần. Trả về chỉ số của phần tử hoặc -1 nếu không tìm thấy. mẫu hàm: def binary_search(arr, left, right, target):

```python
def binary_search(arr, target):
    arr.sort()
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1
arr=list(map(int, input().split()))
target=int(input())
print(binary_search(arr, target))
    
```

```python
def timphantu(n):
    dem={}
    for i in n:
        if i in dem:
            dem[i] += 1
        else:
            dem[i] = 1
    return dem
```

```python
def timphantu(n):
    t=0
    max=0
    for i in n:
        c=n.count(i)
        if c>max:
            max=c
            t=i
    print (t,"xuất hiện nhiều nhất, sớm nhất,",max,"lần")
n=list(map(int, input().split()))
timphantu(n)        
```

Cho danh sách các số
nguyên và một số X, đếm
tất cả các cặp (a, b) sao
cho a + b = X. c

```python
n=list(map(int, input().split()))
x=int(input())
s=0
for i in range(len(n)):
    for j in range(i+1, len(n)):
        for k in range(len(n)):
         if k != i and k != j:
            if n[i] + n[j] == x * n[k]:
                s+=1
                break
print(s)
```
